```python
import pandas as pd
df = pd.read_csv('raw_Superstore.csv')
df.head(10)
```

```python
#gerer les valeurs manquantes / nulles
print(df.isnull().sum())
# supprimer les lignes où Postal Code est manquant
df = df.dropna(subset=['Postal Code'])
```

```python
#gerer les doublons
print(df.duplicated().sum())
df=df.drop_duplicates()
```

```python
# corriger le type de données de la colonne "Order Date" et "Ship Date" (str -> datetime)
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')

# transformer le type de données de la colonne "Postal Code" en entier
df['Postal Code'] = df['Postal Code'].astype(int)
```

```python
#changer les nom de colonnes composées en ajoutant un underscore entre les mots
df.rename(columns={'Row ID': 'Row_ID','Order ID': 'Order_ID','Order Date': 'Order_Date','Ship Date': 'Ship_Date','Ship ID': 'Ship_ID',
                   'Ship Mode': 'Ship_Mode', 'Postal Code': 'Postal_Code','Customer ID': 'Customer_ID','Customer Name': 'Customer_Name',
                   'Product ID': 'Product_ID','Product Name': 'Product_Name'}, inplace=True)
```

```python
df.to_csv('cleaned_Superstore.csv', index=False)
```
